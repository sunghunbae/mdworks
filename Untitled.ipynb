```python
class StageState:
    def __init__(self):
        self.states = []
    
    def add(self, **params):
        """Add a new element, inheriting from the previous element."""
        if self.states:
            new_state = self.states[-1].copy()
            new_state.update(params)
        else:
            params.update({
                'temperature': 10.0, 
                'friction': 1.0,
                'pressure': None,
                'posres_k': 1000.0,
                })
            new_state = params
        self.states.append(new_state)
    
    def __getitem__(self, index):
        return self.states[index]
    
    def __len__(self):
        return len(self.states)
    
    def __repr__(self):
        return f"StageState({self.states})"
```

```python
stages = StageState()
stages.add(temperature=10, friction=5, ps=100)
stages.add(temperature=10, friction=1, ps=24)
stages.add(temperature=300, ps=200)
stages.add(posres_k=0)
stages.add(frequency=50, ps=300)
```

```python
for _ in stages:
    print(_)
```
